### Phase 1: Logic Cơ Bản & Quy đổi Đơn vị (Basic Python)
Xây dựng các hàm tính toán quy đổi giữa Ounce (Quốc tế) và Lượng/Chỉ (Việt Nam) kèm theo tỷ giá hối đoái thực tế.

```python
def calculate_silver_price(spot_usd, usd_vnd_rate, unit_type='tael'):
    # Hằng số chuyển đổi vật lý (1.20565 lượng = 1 oz)
    ratio = 1.20565
    
    # Tính giá VND trên 1 lượng (Tael)
    price_vnd_tael = spot_usd * usd_vnd_rate * ratio
    
    if unit_type == 'chi':
        return price_vnd_tael / 10
    return price_vnd_tael

live_spot = 31.50
current_rate = 25400
print(f"Giá ước tính 1 lượng Bạc tại VN: {calculate_silver_price(live_spot, current_rate):,.0f} VNĐ")
```

### Phase 3: Phân tích Chuyên sâu (Pandas, NumPy & Visualization)
Sử dụng thư viện mạnh mẽ để trích xuất Insights từ bộ dữ liệu 2 năm.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset đã được làm sạch
df = pd.read_csv('silver_cleaned_data.csv')

# Phân tích tương quan giữa giá Thế giới và VN
correlation = df['Global_Price_USD_oz'].corr(df['VN_Spot_Price_VND_Tael'])
print(f"Hệ số tương quan (Correlation): {correlation:.4f}")

# Vẽ biểu đồ Histogram phân phối giá thực tế (Giống với giao diện Web)
plt.figure(figsize=(10, 6))
bins = [0.8e6, 1.0e6, 1.2e6, 1.4e6, 1.6e6, 1.8e6, 2.0e6]
sns.histplot(df['VN_Spot_Price_VND_Tael'], bins=bins, kde=True, color='silver')
plt.title('PHÂN PHỐI MẬT ĐỘ GIÁ BẠC TẠI VIỆT NAM (PHASE 3)')
plt.xlabel('Giá VNĐ/Lượng')
plt.ylabel('Tần suất (Số ngày)')
plt.show()
```
